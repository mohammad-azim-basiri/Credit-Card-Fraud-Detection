```python
import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python

creditcard_dataset = pd.read_csv("../data/creditcard.csv")


print("Dataset Shape:", creditcard_dataset.shape)

print("\nDataset Info:")
creditcard_dataset.info()

display(creditcard_dataset.head())

print("\nDescriptive Statistics:")
display(creditcard_dataset.describe())

print("\nMissing Values:")
display(creditcard_dataset.isnull().sum())

print("\nClass Distribution:")
print(creditcard_dataset["Class"].value_counts())

print(f"class percentage:\n{creditcard_dataset["Class"].value_counts(normalize=True)}")


plt.figure(figsize=(6, 4))
sns.countplot(data=creditcard_dataset, x="Class")
plt.title("Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Transactions")
plt.show()
```

Data quality checks

```python
print("Duplicate rows:", creditcard_dataset.duplicated().sum())

print(creditcard_dataset[creditcard_dataset.duplicated(keep=False)]["Class"].value_counts())
creditcard_dataset = creditcard_dataset.drop_duplicates()

print("Duplicate rows after cleaning:", creditcard_dataset.duplicated().sum())
```

Train/Test split

```python
X = creditcard_dataset.drop("Class",axis=1)
y = creditcard_dataset["Class"]


X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,stratify=y,random_state=42)
print(f"X_train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_test shape: {y_test.shape}")

print("\nTraining class distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTest class distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

standardization

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python

train_test_data = {
    "X_train": X_train_scaled,
    "X_test": X_test_scaled,
    "y_train": y_train,
    "y_test": y_test
}

joblib.dump(train_test_data, "train_test_data.pkl")
```

```python

train_test_data_not_scaled = {
    "X_train": X_train,
    "X_test": X_test,
    "y_train": y_train,
    "y_test": y_test
}

joblib.dump(train_test_data_not_scaled, "train_test_data_not_scaled.pkl")
```
